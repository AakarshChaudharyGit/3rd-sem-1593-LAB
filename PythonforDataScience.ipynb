#SET
## unordered collection  of items
## mutable
## allows the different data types
## it doesn't support for index
## it gives output with uniques (No duplication)
#symbol for set { }

```python
#symbol for set { }
#create a set with similar data items
marks={75,68,89,78,55,75,66,55,99}
print(marks)
```

```python
#create a empty set
a={}
a=set()
#check its type
print(type(a))
```

```python
#set doesn't support the index
students = {"Amit","Riya","Priya"}
print(students[1])
```

```python
#add an item in the set
student={"Amit","Kiran","Sneha"}
student.add("Priya")
print(student)


#addmultiple items or values
student.update(["Ram","Chitra","Vani","Devi"])
print(student)
```

```python
#remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Riya")
print(students)

students.discard("Amit")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.remove("Karan")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.discard("Karan")
print(students)
```

```python
#create a set and clear list of item
students_info={"Amit","Riya","Priya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99.88,77,66}
del marks
print(marks)
```

```python
#set operations
python_students={"Amit","Neha","Sanjiv","Manish"}
java_students={"Amit","Karan","Dillip","Manish","Neha"}

#for union operation by using set
result=python_students.union(java_students)
print("Union operation:",result)

# for intersection operation by using set
result_intersection=python_students.intersection(java_students)
print("Intersection operation:",result_intersection)

#for difference operation by using set
result_difference=python_students.difference(java_students)
print("Difference operation:", result_difference)

```

```python
# convert list to set
stud_id=[101,102,103,202,103,104,105]
unique_stud_id = set(stud_id)
print(unique_stud_id)
unique_roll_n0 =list(unique_stud_id)
print(unique_roll_n0)
```
